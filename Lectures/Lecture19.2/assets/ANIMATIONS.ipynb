Bessa's circular top and bottom rings replaced by ellipses, with a phase offset between their major axes.

**Breaking the rings' symmetry destroys coiling.** 9 of 67 designs coiled at all, and none were feasible.

both campaigns the single best compressive strain ever reached was 0.78, against the 0.80 required,
and even that design failed other criteria.

A later 32-point sweep returned coilable=0 at every point, including two probes perturbed about one
degree from circular.

## The rectangular section

<p align="center"><img src="../../figures/D6_run17_rectangle_native.gif" width="40%"></p>

An anisotropic rectangular longeron, short in the radial direction and long in the tangential one.

**The one that worked.** 6 feasible of 165, and the baseline every later idea is measured against.

It becomes run17_rectangle, the study's canonical anchor, later refined to 0.7704 kPa.

Peak stress scales as E·w·c³/L². The depth c is capped by how tightly the design must coil, the width
w by the slenderness >= 10 beam-validity gate. A circle is isotropic, one parameter, so any circle
stiff enough to carry load is too stiff to coil within the 2% strain budget. A rectangle separates the
two demands.

Idea D6, verdict SUPPORTED and WORKS. Funnel n=165 to 149 coilable to 148 Riks-solved to 6 feasible.
No stress history exists for this family in the deck.

## Legged snap-chain

<div style="display:flex; align-items:center; justify-content:center; gap:3%; margin-top:0.5em">
  <img src="../../figures/snaplegs_C1_winner.gif" style="width:52%">
  <img src="../../figures/snaplegs_C1_winner_mini.png" style="width:41%">
</div>

Rigid rings joined by curved arches, each on two short legs so it can flip to its mirror curve without hitting the ring below. Compression comes from the arches snapping in sequence.

**It abandons ring-rotation coiling entirely, and reaches 10.79x the Bessa point.**

Grounded in Shan et al. 2015 on chained bistable curved-beam units, with the bistability threshold
from Qiu, Lang and Slocum 2004.

The best design verifies exactly against the run store: sigma_peak 1.2104 kPa, compressive strain
0.8204, local strain 0.0188, converged. All three feasibility criteria met, and 1.2104 / 0.1122 is
10.79x Bessa. Its stress peaks smoothly rather than spiking.

Stage 1 was deliberately skipped: a finite-deformation snap-through of an already-curved member is
not a small perturbation about an undeformed state, so a linear buckling check would describe global
sway of the ring stack, not this mechanism.

Verification is the strongest in the study: sigma_peak flat to six significant figures across a 100x
finer arc-length and unchanged under 2x mesh refinement, three separately converged designs clearing
the target, no non-adjacent self-interference, and 99.64% height recovery on unload from 82%
compression, with the deck itself noting the last 0.36% sits at the noise floor.

Two things to carry. All three runs closed on an external API quota limit rather than on the science
finishing, with five hypotheses queued and untested. And the deck's stats line reads 86 feasible; the
three run stores contain 19. The n=168 reconciles exactly, so the denominator is right and that
numerator needs checking before it goes on a slide.

## Tape-spring longerons

<div style="display:flex; align-items:center; justify-content:center; gap:3%; margin-top:0.5em">
  <img src="../../figures/D25_tape_spring_landscape.gif" style="width:52%">
  <img src="../../figures/D25_tape_spring_mini.png" style="width:41%">
</div>

The solid beam replaced by a thin-walled open circular arc, a tape-spring section, on the hypothesis that a localized elastic fold could escape the mast-scale curvature limit.

**The largest campaign in the study, 406 designs, and none feasible.**

The curvature that keeps the arc locally stable is the same curvature that sets its bending-strain
floor. It can never be shallow enough to fold locally without buckling first, so strain follows
ordinary beam bending at every geometry tested. A steel tape measure snaps the same way.

Idea D25, verdict POWERED and REFUTED. Funnel n=406 to 176 coilable to 137 Riks-solved to 0
feasible.

## Tensegrity longerons

<p align="center"><img src="../../figures/D21_tensegrity_native.gif" width="40%"></p>

The bending longeron replaced by a pin-jointed, prestressed strut-and-cable assembly, taking its stiffness from prestress and geometry rather than from bending.

**The largest stress in the entire study, and disqualified for it.**

Colour is LE11, axial strain. There is no beam bending field to show here, which is the point.

Maximum local strain came out near 9e-14. The members do not strain: compression is achieved by rigid
bars rotating about joints, so the object is a linkage and its buckling load is not on the same
footing as elastic bending. The problem statement's apples-to-apples clause was written after this
design exploited its absence.

The critic rejected the run on a second ground: the 287x headline contradicted the run's own
commissioned literature review, which had predicted a modest saturating gain, and the gap was never
reconciled.

Idea D21, verdict SUPPORTED (DISQUALIFIED), DEAD-END. Funnel n=45 to 45 coilable to 44 Riks-solved to
12 feasible.

## The re-study under contact

<p align="center"><img src="../../figures/restudy_floor_contact.gif" width="50%"></p>

Ground and disc contact were restored to the simulator on 2026-08-06, and five earlier families were re-examined under the new physics.

**One of the five was settled. Three were never tested.**

Tape-spring was settled: 330 designs over two campaigns, and the best of 28 decided reaches 21%
compression against the 80% required. Tensegrity was stopped by the floor, energy up 86% with stress
unchanged, still blocked on printability. Kresling, laced and chiral-shell were migrated to the new
oracle but never tested under it.

From the deck: do not read "we migrated it" as "we tested it". One design cannot settle a family, and
the one design available is usually the winner of a search run carried out without contact.
