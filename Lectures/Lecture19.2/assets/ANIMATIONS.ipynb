Bessa's circular top and bottom rings replaced by ellipses, with a phase offset between their major axes.

**Breaking the rings' symmetry destroys coiling.** 9 of 67 designs coiled at all, and none were feasible.

The chart is the reason it is a dead end rather than an open question: across all 67 evaluations in
both campaigns the single best compressive strain ever reached was 0.78, against the 0.80 required,
and even that design failed other criteria.

A later 32-point sweep returned coilable=0 at every point, including two probes perturbed about one
degree from circular.

Idea D10. Funnel n=67 to 9 coilable to 9 Riks-solved to 0 feasible. Verdict INCONCLUSIVE, DEAD-END.

## The rectangular section

<div style="display:flex; align-items:center; justify-content:center; gap:3%; margin-top:0.5em">
  <img src="../../figures/D6_run17_rectangle_native.gif" style="width:44%">
  <img src="../../figures/D6_run17_rectangle_mini.png" style="width:48%">
</div>

An anisotropic rectangular longeron, short in the radial direction and long in the tangential one.

**The one that worked.** 6 feasible of 165, and the baseline every later idea is measured against.

It becomes run17_rectangle, the study's canonical anchor, later refined to 0.7704 kPa.

The curve stays red for its whole history, which is the point: local strain never breaches the 2%
ceiling, so nothing greys out. It peaks at 5.41x Bessa near 6% compression and decays smoothly
afterwards, the textbook post-buckling signature.

Peak stress scales as E·w·c³/L². The depth c is capped by how tightly the design must coil, the
width w by the slenderness >= 10 beam-validity gate. A circle is isotropic, one parameter, so any
circle stiff enough to carry load is too stiff to coil within the strain budget. A rectangle
separates the two demands.

Idea D6, verdict SUPPORTED and WORKS. Funnel n=165 to 149 coilable to 148 Riks-solved to 6
feasible. The chart was generated from this design's own archived Riks solve using the study's own
bo/mini_chart.py, and reports sigma_peak 0.60713 kPa, matching the study's recorded value.

## Pre-twisted longerons

<div style="display:flex; align-items:center; justify-content:center; gap:3%; margin-top:0.5em">
  <img src="../../figures/D1_pretwisted_strongest.gif" style="width:46%">
  <img src="../../figures/D1_pretwisted_strongest_mini.png" style="width:47%">
</div>

A helical pre-twist applied to each leg, on top of the full 7-parameter Bessa cross-section search.

**Nothing here was ever checked against the strain limit.** Not the twisted designs, and not the untwisted baseline the study called its floor.

Grey here means unmeasured, not failed. No design in this family ever reached a second-stage
solve, so its numbers are Stage-1 eigenvalue estimates never tested against the 2% cap. The 195x
is normalised against the circular family's 0.1122 kPa, not this family's own 65.31 kPa baseline.

It cannot be measured as modelled: Abaqus rejects a generalized cross-section paired with
DURING_ANALYSIS integration, and the mode it does accept writes no strain field. The elliptical
family substituted a rectangle sized to match, which works for two targets. This box has four.

INCONCLUSIVE rather than FALSIFIED for a separate reason, the licence server: the criterion
required 80 evaluations, 26 Abaqus runs died to licence-server errors, and the validator caught
the strategizer closing on 46.

Idea D1. Funnel n=46 to 6 coilable to 0 Riks-solved to 0 feasible.

## Legged snap-chain

<div style="display:flex; align-items:center; justify-content:center; gap:3%; margin-top:0.5em">
  <img src="../../figures/snaplegs_C1_winner.gif" style="width:52%">
  <img src="../../figures/snaplegs_C1_winner_mini.png" style="width:41%">
</div>

Rigid rings joined by curved arches, each on two short legs so it can flip to its mirror curve without hitting the ring below. Compression comes from the arches snapping in sequence.

**It abandons ring-rotation coiling entirely, and reaches 10.79x the Bessa point.**

Grounded in Shan et al. 2015 on chained bistable curved-beam units, with the bistability threshold
from Qiu, Lang and Slocum 2004.

The best design verifies exactly against the run store: sigma_peak 1.2104 kPa, compressive strain
0.8204, local strain 0.0188, converged. All three feasibility criteria met, and 1.2104 / 0.1122 is
10.79x Bessa. Its stress peaks smoothly rather than spiking.

Stage 1 was deliberately skipped: a finite-deformation snap-through of an already-curved member is
not a small perturbation about an undeformed state, so a linear buckling check would describe global
sway of the ring stack, not this mechanism.

Verification is the strongest in the study: sigma_peak flat to six significant figures across a 100x
finer arc-length and unchanged under 2x mesh refinement, three separately converged designs clearing
the target, no non-adjacent self-interference, and 99.64% height recovery on unload from 82%
compression, with the deck itself noting the last 0.36% sits at the noise floor.

Two things to carry. All three runs closed on an external API quota limit rather than on the science
finishing, with five hypotheses queued and untested. And the deck's stats line reads 86 feasible; the
three run stores contain 19. The n=168 reconciles exactly, so the denominator is right and that
numerator needs checking before it goes on a slide.

## Tape-spring longerons

<div style="display:flex; align-items:center; justify-content:center; gap:3%; margin-top:0.5em">
  <img src="../../figures/D25_designC_contact_riks_landscape.gif" style="width:46%">
  <img src="../../figures/D25_designC_contact_riks_mini.png" style="width:47%">
</div>

The solid beam replaced by a thin-walled open circular arc, a tape-spring section, on the hypothesis that a localized elastic fold could escape the mast-scale curvature limit.

**It reaches 21% compression before breaking the strain limit. It needs 80%.**

Read the colour. The curve is red while local strain is inside the 2% ceiling and greys out the
moment it is not. It greys at 0.21 compression, so everything from there to 0.80 is a design that
has already failed. The peak, 2.85x Bessa, arrives at about 1.5% compression and decays from there.

The mechanism defeats itself: the curvature that keeps the arc locally stable is the same curvature
that sets its bending-strain floor. It can never be shallow enough to fold locally without buckling
first, so strain follows ordinary beam bending at every geometry tested. A steel tape measure snaps
the same way.

Shown here is Design C, re-solved under contact. The original campaign was the largest in the
study, 406 designs and none feasible. The contact re-study added 330 more designs over two
campaigns and settled it.

Idea D25, verdict POWERED and REFUTED, mechanism localized elastic fold.

## Tensegrity longerons

<div style="display:flex; align-items:center; justify-content:center; gap:3%; margin-top:0.5em">
  <img src="../../figures/D21-2_restudy_tensegrity_contact.gif" style="width:52%">
  <img src="../../figures/D21-2_tensegrity_strain_history_mini.png" style="width:41%">
</div>

The bending longeron replaced by a pin-jointed, prestressed strut-and-cable assembly, taking its stiffness from prestress and geometry rather than from bending.

**It posted the largest stress in the study, and was disqualified for it.**

Shown here under contact, re-tested after ground contact was restored study-wide on 2026-08-06.
Contact had to be added by hand for this family: the struts carry no cross-section geometry, so the
secondary surface is a node region with the strut radius injected explicitly.

The original result: maximum local strain near 9e-14. The members did not strain at all. Compression
was achieved by rigid bars rotating about joints, so the object is a linkage and its buckling load is
not on the same footing as elastic bending. The problem statement's apples-to-apples clause was
written after this design exploited its absence.

Under contact the picture changes and still does not rescue it. One design, tested on and off:
energy absorbed rises 86%, from 18.202 to 33.780 kPa, and genuine straining appears where there was
none, roughly 0 to 3.4%. But 3.4% is above the 2% ceiling, sigma_peak is bit-identical because the
peak lands before contact engages, and the mechanism is still a linkage needing pin joints and
prestressed cables that will not survive printing.

Idea D21, verdict SUPPORTED (DISQUALIFIED) and DEAD-END; the contact re-test is D21-2, POWERED and
REFUTED.

## The re-study under contact

<p align="center"><img src="../../figures/restudy_floor_contact.gif" width="50%"></p>

Ground and disc contact were restored to the simulator on 2026-08-06, and five earlier families were re-examined under the new physics.

**One of the five was settled. Three were never tested.**

Tape-spring was settled: 330 designs over two campaigns, and the best of 28 decided reaches 21%
compression against the 80% required. Tensegrity was stopped by the floor, energy up 86% with stress
unchanged, still blocked on printability. Kresling, laced and chiral-shell were migrated to the new
oracle but never tested under it.

From the deck: do not read "we migrated it" as "we tested it". One design cannot settle a family, and
the one design available is usually the winner of a search run carried out without contact.
